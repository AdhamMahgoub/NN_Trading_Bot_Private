```python
import sys, datetime
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if not (Path.cwd() / 'A_Data_Gathering').exists() else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from A_Data_Gathering.helpers import load_tickers, check_and_refresh_date_config, download_tickers
```

```python
# ── Data config ──────────────────────────────────────────────────────────────
SEED           = 42
INTERVAL       = '1d'
START_DATE     = pd.Timestamp('2025-01-01')
END_DATE       = pd.Timestamp(datetime.date.today())
TRAIN_END_DATE = pd.Timestamp('2025-11-30')
VAL_END_DATE   = pd.Timestamp('2026-01-20')
TICKER_SUBSET  = None   # None → all tickers | ['AAPL', 'MSFT'] → specific symbols | 42 → random sample of that size
```

```python
RUN_START_TIME = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
RUN_OUTPUT_DIR = PROJECT_ROOT / 'D_Models' / 'run_output' / RUN_START_TIME
RUN_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f'Run output: {RUN_OUTPUT_DIR}')
```

```python
TICKERS = load_tickers(PROJECT_ROOT / 'A_Data_Gathering' / 'tickers.txt',
                       subset=TICKER_SUBSET, seed=SEED)
print(f'Using {len(TICKERS)} tickers')
```

```python
STOCKS_DIR   = PROJECT_ROOT / 'A_Data_Gathering' / 'dataset' / 'stocks'
CONFIG_PATH  = PROJECT_ROOT / 'A_Data_Gathering' / 'date_config.txt'

check_and_refresh_date_config(
    current_config={
        'INTERVAL':       str(INTERVAL),
        'START_DATE':     str(START_DATE.date()),
        'END_DATE':       str(END_DATE.date()),
        'TRAIN_END_DATE': str(TRAIN_END_DATE.date()),
        'VAL_END_DATE':   str(VAL_END_DATE.date()),
        'TICKER_SUBSET':  str(TICKER_SUBSET),
    },
    config_path=CONFIG_PATH,
    stocks_dir=STOCKS_DIR,
)
```

```python
_summary = download_tickers(
    tickers=TICKERS, start=START_DATE, end=END_DATE,
    interval=INTERVAL, out_dir=STOCKS_DIR,
)
```
